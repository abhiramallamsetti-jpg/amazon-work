# Dataset Exploration - Day 1

This notebook performs basic exploratory analysis on the Customer Support on Twitter dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set styling for plots
sns.set_theme(style="whitegrid")
```

## Load Dataset

```python
df = pd.read_csv('../data/raw/twcs/twcs.csv')
print("Dataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
```

## Basic Information

```python
print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

## Analysis

```python
# Brands are typically authors of inbound=False tweets
brands = df[df['inbound'] == False]['author_id'].unique()
print(f"Number of unique brands: {len(brands)}")

# Customers are typically authors of inbound=True tweets
customers = df[df['inbound'] == True]['author_id'].unique()
print(f"Number of unique customers: {len(customers)}")

print(f"Total Number of tweets: {len(df)}")
```

```python
df['created_at'] = pd.to_datetime(df['created_at'], errors='coerce')
print(f"Date Range: {df['created_at'].min()} to {df['created_at'].max()}")
```

## Top Brands

```python
top_brands = df[df['inbound'] == False]['author_id'].value_counts().head(10)
print("Most frequent brands:")
print(top_brands)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x=top_brands.values, y=top_brands.index)
plt.title('Top 10 Brands by Number of Replies')
plt.xlabel('Number of Tweets')
plt.ylabel('Brand')
plt.tight_layout()
plt.show()
```

## Customer vs Brand Replies

```python
inbound_counts = df['inbound'].value_counts()
plt.figure(figsize=(6, 4))
sns.barplot(x=['Customer (True)', 'Brand (False)'], y=inbound_counts.values)
plt.title('Customer vs Brand Tweets')
plt.ylabel('Count')
plt.show()
```

## Text Length Statistics

```python
df['text_length'] = df['text'].str.len()
print(df['text_length'].describe())

plt.figure(figsize=(10, 6))
sns.histplot(df['text_length'], bins=50, kde=False)
plt.title('Distribution of Tweet Lengths')
plt.xlabel('Length of Tweet (Characters)')
plt.ylabel('Frequency')
plt.show()
```

## Example Tweets

```python
print("Example Brand Tweets:\n")
for brand in top_brands.index[:3]:
    print(f"--- {brand} ---")
    sample_tweets = df[(df['author_id'] == brand) & (df['inbound'] == False)]['text'].head(3)
    for idx, text in enumerate(sample_tweets):
        print(f"{idx+1}. {text}")
    print("\n")
```
