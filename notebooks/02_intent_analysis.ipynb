# Intent Analysis - Day 2

This notebook analyzes the discovered intents from the SpotifyCares support cases.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## Load Data

```python
cases = pd.read_csv('../data/processed/support_cases.csv')
intents = pd.read_csv('../data/processed/intent_distribution.csv')
print(f"Number of support cases: {len(cases)}")
```

## Intent Distribution

```python
print(intents)

plt.figure(figsize=(10, 6))
sns.barplot(data=intents, x='percentage', y='intent')
plt.title('Distribution of Customer Intents')
plt.xlabel('Percentage of Cases (%)')
plt.ylabel('Intent')
plt.show()
```

## Example Messages

```python
examples = pd.read_csv('../data/processed/intent_examples.csv')
for intent in examples['intent'].unique()[:3]:
    print(f"\n--- {intent.upper()} ---")
    sample = examples[examples['intent'] == intent].head(2)
    for _, row in sample.iterrows():
        print(f"Customer: {row['customer_message']}")
        print(f"Brand: {row['brand_reply']}\n")
```
